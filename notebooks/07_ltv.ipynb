# 7. Сценарный LTV по сегментам

Расчёт использует среднюю историческую выручку клиента и явные предположения о churn, горизонте, ставке дисконтирования и CAC. Это revenue LTV, а не маржинальный LTV, поэтому результат нельзя напрямую считать прибылью.


```python
from pathlib import Path

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
ARTIFACTS = ROOT / 'artifacts'
ARTIFACTS.mkdir(exist_ok=True)
```

```python
import pandas as pd
from ecommerce_analysis.ltv import calculate_segment_ltv

rfm = pd.read_csv(ARTIFACTS / 'rfm_customers.csv', index_col=0)
ltv = calculate_segment_ltv(
    rfm,
    annual_churn_rate=0.20,
    discount_rate=0.15,
    horizon_years=5,
    cac=2000,
)
ltv
```
